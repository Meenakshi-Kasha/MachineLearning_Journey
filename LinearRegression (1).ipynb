b. Train Linear Regression

```python
from sklearn.linear_model import LinearRegression
import pandas as pd

df = pd.DataFrame({
    "Hours":[1,2,3,4,5],
    "Marks":[10,20,30,40,50]
})

x = df[['Hours']]
y = df['Marks']

model = LinearRegression()
model.fit(x, y)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

Practical 4: Visualize Linear Regression

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
df = pd.DataFrame({
    "Hours":[1,2,3,4,5],
    "Marks":[10,20,30,40,50]
})

x = df[['Hours']]
y = df['Marks']

model = LinearRegression()
model.fit(x, y)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)

plt.scatter(x,y)
plt.plot(x,model.predict(x))

plt.xlabel("Hours")
plt.ylabel("marks")
plt.show()
```

Practical 5: Real Dataset (Recommended)

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
data= load_diabetes()
x= data.data
y= data.target
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)

model = LinearRegression()

model.fit(x_train,y_train)

score = model.score(x_test,y_test)

print("R2 Score:", score)
```

Practical 6: Understanding Overfitting Intuition

Create polynomial regression.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,4,5,6,7,8])
y = np.array([2,4,6,8,10,12,14,16])

coeff = np.polyfit(x,y,7)

poly = np.poly1d(coeff)

plt.scatter(x,y)
plt.plot(x,poly(x))

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,4,5,6,7,8])
y = np.array([2,5,4,9,10,13,12,17])  # noisy data

coeff = np.polyfit(x, y, 7)

poly = np.poly1d(coeff)

x_new = np.linspace(1,8,100)

plt.scatter(x,y)
plt.plot(x_new, poly(x_new))

plt.show()
```
